To visually inspect outliers, I’m using boxplots for each numeric field. Boxplots make it easier to see extreme values beyond the typical range.

Outliers in this dataset can be valid (example: very large online universities with unusually high enrollment, or elite private schools with unusually high tuition, or medical/tech schools with unusually high earnings). However, outliers can also reflect data limitations such as:

Finally, I’m pulling the actual rows associated with extreme outliers using the IQR method. This helps answer: “What does the outlier represent?”

In this dataset, outliers are often real institutions with legitimate extreme values (example: very large enrollment schools, high-tuition private colleges, or high-earning specialized programs). These are typically valid data, not “dirty” data—but they can strongly influence averages and correlations, so I may need to use medians, log scales, or filtering later.

```python
def iqr_bounds(series):
    s = series.dropna()
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - 1.5*iqr, q3 + 1.5*iqr

outlier_fields = ["UGDS", "TUITIONFEE_IN", "TUITIONFEE_OUT", "DEBT_MDN_NUM", "MD_EARN_WNE_P10"]

for col in outlier_fields:
    low, high = iqr_bounds(df_use[col])
    out = df_use.loc[df_use[col].notna() & (df_use[col] > high), ["INSTNM", "CONTROL", col]].sort_values(col, ascending=False).head(10)
    
    print(f"\n--- Outliers (high end) for {col} ---")
    print(f"IQR upper bound ≈ {high:,.2f}")
    display(out)
```

In this step, I am preparing the College Scorecard dataset for analysis. The goal is to ensure the fields I plan to use are correctly typed, consistently formatted, and free of issues that would distort the EDA (such as duplicates, non-numeric codes, or extreme invalid values).

Before each change, I will check the data “before and after” using shape, info(), and sample rows to confirm the code worked as expected. I will also avoid dropping missing values too early so I don’t accidentally remove useful data.

```python
import pandas as pd
import numpy as np

file_path = "Most-Recent-Cohorts-Institution.csv"

df = pd.read_csv(file_path, low_memory=False)

use_cols = [
    "INSTNM", "CONTROL", "UGDS", "TUITIONFEE_IN", "TUITIONFEE_OUT",
    "C150_4", "DEBT_MDN", "MD_EARN_WNE_P10"
]

df_use = df[use_cols].copy()

print("BEFORE cleaning shape:", df_use.shape)
df_use.head()
```

Dropping unused columns

The original Scorecard dataset contains many columns that are not needed for my Phase 1 EDA goals. To keep the dataset focused and easier to manage, I am keeping only the variables I plan to analyze and dropping all other fields. This also reduces noise and prevents accidental use of irrelevant variables later.

```python
df_use.info()
```

Checking for duplicates

Since each row should represent one institution, duplicate rows would inflate counts and distort summaries. I am checking for duplicates across all retained columns and removing them if found.

```python
before = df_use.shape[0]

for col in ["UGDS", "TUITIONFEE_IN", "MD_EARN_WNE_P10"]:
    cutoff = df_use[col].quantile(0.99)
    df_use = df_use[df_use[col].isna() | (df_use[col] <= cutoff)]

after = df_use.shape[0]
print("Rows before trimming:", before)
print("Rows after trimming:", after)
```

Converting DEBT_MDN to numeric

During inspection, DEBT_MDN included non-numeric codes (such as privacy suppression values). Since my analysis will treat debt as a numerical variable, I am creating a cleaned numeric version of this column. Non-numeric values will be converted to missing (NaN) so they are handled consistently later.

```python
print("BEFORE conversion - DEBT_MDN sample values:")
print(df_use["DEBT_MDN"].value_counts(dropna=False).head(10))

df_use["DEBT_MDN_NUM"] = pd.to_numeric(df_use["DEBT_MDN"], errors="coerce")

print("\nAFTER conversion - DEBT_MDN_NUM describe:")
df_use["DEBT_MDN_NUM"].describe()
```

Outliers: identify “dirty” vs. legitimate

The Scorecard dataset naturally contains legitimate outliers (for example, very large universities or very expensive private schools). Those represent real institutions and should usually be kept.

However, outliers that are invalid (examples: negative tuition, negative enrollment, completion rates outside 0–1) represent dirty data. In this step, I will remove rows with clearly invalid values, while keeping legitimate high-end values that reflect real-world differences.

```python
print("BEFORE invalid-value filtering shape:", df_use.shape)

# Define validity rules
valid_mask = (
    (df_use["UGDS"].isna() | (df_use["UGDS"] >= 0)) &
    (df_use["TUITIONFEE_IN"].isna() | (df_use["TUITIONFEE_IN"] >= 0)) &
    (df_use["TUITIONFEE_OUT"].isna() | (df_use["TUITIONFEE_OUT"] >= 0)) &
    (df_use["C150_4"].isna() | ((df_use["C150_4"] >= 0) & (df_use["C150_4"] <= 1))) &
    (df_use["MD_EARN_WNE_P10"].isna() | (df_use["MD_EARN_WNE_P10"] >= 0)) &
    (df_use["DEBT_MDN_NUM"].isna() | (df_use["DEBT_MDN_NUM"] >= 0))
)

invalid_rows = (~valid_mask).sum()
print("Invalid rows to remove:", invalid_rows)

df_use = df_use.loc[valid_mask].copy()

print("AFTER invalid-value filtering shape:", df_use.shape)
df_use.head()
```

New calculations

To support analysis, I am creating a new calculated field: an estimate of the tuition gap between out-of-state and in-state tuition (TUITION_GAP). This helps quantify how much more expensive a school is for non-resident students and may reveal patterns by institution type.

```python
df_use["TUITION_GAP"] = df_use["TUITIONFEE_OUT"] - df_use["TUITIONFEE_IN"]

# Inspect the new field
df_use[["TUITIONFEE_IN", "TUITIONFEE_OUT", "TUITION_GAP"]].describe()
df_use[["INSTNM", "TUITIONFEE_IN", "TUITIONFEE_OUT", "TUITION_GAP"]].head()
```

Binning/grouping

To make comparisons easier, I am binning institutions into enrollment size categories based on undergraduate enrollment (UGDS). These categories will support comparisons of tuition, earnings, and completion rates across different-sized schools.

```python
bins = [-1, 999, 4999, 14999, np.inf]
labels = ["Small (<1k)", "Medium (1k–4.9k)", "Large (5k–14.9k)", "Very Large (15k+)"]

df_use["ENROLL_BIN"] = pd.cut(df_use["UGDS"], bins=bins, labels=labels)

df_use["ENROLL_BIN"].value_counts(dropna=False)
```

Merging/concatenation

For this Phase 1 dataset, I am not merging in additional datasets. The College Scorecard file already contains the core cost and outcome fields needed for my EDA goals. If I later add geographic or ranking data, I will merge using a stable identifier and confirm the join worked by checking row counts before and after.

Dropping missing values (last step)

I am dropping missing values at the end to avoid losing potentially useful data too early. Since some fields are missing due to reporting rules or privacy suppression, I will only drop rows that are missing values in the key analysis variables needed for my planned comparisons.

For this EDA, the key outcome fields are: earnings, tuition (in-state), enrollment, and completion rate. Dropping nulls here creates a consistent dataset for analysis without removing rows prematurely.

```python
print("BEFORE dropna shape:", df_use.shape)

key_fields = ["UGDS", "TUITIONFEE_IN", "C150_4", "MD_EARN_WNE_P10"]
df_clean = df_use.dropna(subset=key_fields).copy()

print("AFTER dropna shape:", df_clean.shape)

df_clean.info()
df_clean.head()
```

**Purpose:** Load the cleaned dataset, create readable institution-type labels, and generate initial summary checks/visuals to establish a baseline view of earnings patterns by institution type.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the clean dataset from Phase 1
df = pd.read_csv('college_scorecard_clean.csv')

# Create readable CONTROL label
control_map = {1: 'Public', 2: 'Private Nonprofit', 3: 'Private For-Profit'}
df['CONTROL_LABEL'] = df['CONTROL'].map(control_map)

print(f"Dataset shape: {df.shape}")
print("\nInstitution types:")
print(df['CONTROL_LABEL'].value_counts())
print("\nEnrollment size distribution:")
print(df['ENROLL_BIN'].value_counts())
```

## 2. Initial Exploration – Summary Statistics

**Purpose:** Compute descriptive statistics for the main variables to understand their central tendency, spread, and range before testing relationships.

```python
key_vars = ['C150_4', 'TUITIONFEE_IN', 'DEBT_MDN_NUM', 'MD_EARN_WNE_P10']
print(df[key_vars].describe().round(0))

# Quick look at earnings-to-debt ratio (only where debt > 0)
df['EARNINGS_TO_DEBT'] = df['MD_EARN_WNE_P10'] / df['DEBT_MDN_NUM']
print("\nEarnings-to-Debt Ratio (where debt available):")
print(df['EARNINGS_TO_DEBT'].describe().round(2))
```

### Interpretation

Median earnings 10 years after entry (`MD_EARN_WNE_P10`) average approximately **51,135**, with a standard deviation of about **13,760** and a range extending from **13,845** to **138,000**. This shows that long-term economic outcomes also vary widely across institutions.

## 3. Hypothesis 1: Completion Rate vs. Long-Term Earnings

**Purpose:** Evaluate whether higher 4-year completion rates are associated with higher 10-year median earnings using both correlation and a scatterplot trendline.

**Question / Hypothesis** Schools with higher graduation rates will tend to have higher median earnings 10 years after entry (moderate positive relationship expected).

```python
# Correlation
corr_comp = df[['C150_4', 'MD_EARN_WNE_P10']].corr().iloc[0,1]
print(f"Pearson correlation (Completion Rate ↔ Earnings): {corr_comp:.3f}")

# Scatter plot with regression line
plt.figure(figsize=(10, 6))
sns.regplot(
    data=df,
    x='C150_4',
    y='MD_EARN_WNE_P10',
    scatter_kws={'alpha':0.25, 's':35},
    line_kws={'color':'darkred', 'lw':2.5}
)
plt.title('Graduation Rate vs Median Earnings 10 Years After Entry', fontsize=13)
plt.xlabel('4-Year Completion Rate (C150_4)')
plt.ylabel('Median Earnings 10 Years After Entry ($)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretation

The correlation and scatterplot indicate whether completion rate and earnings move together and whether the relationship appears weak, moderate, or strong.

The Pearson correlation between 4-year completion rate (`C150_4`) and median earnings 10 years after entry (`MD_EARN_WNE_P10`) is approximately **0.586**. This indicates a **moderately strong positive relationship**: institutions that graduate a larger proportion of their students tend to have graduates with noticeably higher long-term earnings.

The regression line in the scatter plot clearly slopes upward, showing that as completion rates increase from near 0% to near 100%, predicted median earnings rise from roughly **30,000–35,000** to **60,000–70,000** or more.

However, there is still considerable scatter around the line. Many schools with similar graduation rates produce quite different earnings outcomes. For example, institutions in the **0.4–0.6 completion rate range** show earnings outcomes ranging from approximately **35,000** to over **80,000**.

This suggests that while completion rate is one of the more meaningful single predictors examined so far, it is **not sufficient on its own**. Other factors likely explain a substantial portion of the remaining variation in earnings outcomes, including:

- Institutional selectivity
- Academic program mix (e.g., STEM vs. non-STEM emphasis)
- Geographic region and local labor market conditions
- Student body characteristics

Overall, completion rate appears to be an important indicator, but it should be interpreted as part of a broader institutional and economic context rather than as a standalone determinant of earnings.

## 4. Hypothesis 2: Tuition Cost vs. Earnings
**Purpose:** Test whether in-state tuition is associated with long-term earnings outcomes by quantifying the correlation and visualizing the relationship.

**Question / Hypothesis** Higher tuition will be associated with higher earnings, but the relationship may be weaker than completion rate because tuition often proxies for institutional prestige rather than directly causing better outcomes.

```python
corr_tuit = df[['TUITIONFEE_IN', 'MD_EARN_WNE_P10']].corr().iloc[0,1]
print(f"Pearson correlation (In-State Tuition ↔ Earnings): {corr_tuit:.3f}")

plt.figure(figsize=(10, 6))
sns.regplot(
    data=df,
    x='TUITIONFEE_IN',
    y='MD_EARN_WNE_P10',
    scatter_kws={'alpha':0.20, 's':30},
    line_kws={'color':'darkred'}
)
plt.title('In-State Tuition vs Median Earnings 10 Years After Entry', fontsize=13)
plt.xlabel('In-State Tuition ($)')
plt.ylabel('Median Earnings 10 Years After Entry ($)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretation

These results show whether higher tuition is associated with higher earnings, while also revealing whether the pattern looks consistent or mixed across institutions.

The Pearson correlation between in-state tuition (`TUITIONFEE_IN`) and median earnings 10 years after entry (`MD_EARN_WNE_P10`) is approximately **0.468**, indicating a **moderate positive relationship**.

This suggests that institutions with higher in-state tuition tend to report higher median earnings on average. The upward-sloping regression line in the scatter plot supports this pattern, with predicted earnings generally increasing as tuition rises.

However, the relationship is **not especially strong**, and there is substantial variation around the trend line. At many tuition levels, schools show a wide spread of earnings outcomes. For example, some lower- to mid-tuition institutions produce earnings comparable to (or even higher than) more expensive schools.

This means tuition alone is an **imperfect predictor** of long-term earnings. In-state tuition may be acting as a proxy for other institutional characteristics, such as:

- Selectivity/admissions standards  
- Program mix (e.g., STEM, business, health fields)  
- Institution type (public vs. private nonprofit vs. for-profit)  
- Geographic region and local labor markets  
- Student population characteristics  

Overall, the pattern suggests that higher tuition is **associated** with higher earnings, but the effect is **moderate and noisy**, so tuition should not be interpreted as a standalone driver of earnings outcomes.

## 5. Hypothesis 3: Median Debt vs. Earnings

- Tuition levels and pricing structure

Overall, the pattern suggests that median debt is **moderately associated** with earnings, but it is a **noisy and indirect indicator**. Debt should be analyzed alongside completion rates, tuition, and institution type to better understand long-term earnings outcomes.
